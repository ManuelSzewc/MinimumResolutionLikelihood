# 1D Gaussian

```python
ndim = 1
background_model = MyPDF(norm(loc=0,scale=1), ndim=1)
biased_background_model = MyPDF(norm(loc=0,scale=1.1), ndim=1)
signal_model = MyPDF(norm(loc=0.5,scale=0.1), ndim=1)
```

```python
analysis = Analysis(signal_model, background_model, biased_background_model, ndim=ndim,results_folder='results_gaussian_1d')
```

```python
Nevents = int(5e4)#100000
Z = 1.0
B = (Nevents + 0.5*Z**2) * (1 - np.sqrt(1 - Nevents**2 / ((Nevents +0.5* Z**2)**2)))
S = (Nevents-B)
print(S,B)
analysis.run_one_experiment_save_plots(S=S,B=B,CRsize=5*Nevents,Toysize=5*Nevents,seed=42,eps=0.01,density_plot=True)
```

```python
Svals = np.array([10.0,75.0,150.0,300.0])
y_coordinates = [[1,60,20],[1,20,10],[1,20,10],[1,20,10]]
y_coordinates_2 = [[0.71,0.2,0.5],[0.71,0.20,0.5],[0.71,0.20,0.5],[0.71,0.20,0.5]]
analysis.run_Sinj_scan(Svals=Svals,n_experiments=500,B=B,CRsize=5*Nevents,Toysize=5*Nevents,seeds=[i for i in range(len(Svals))],eps=0.0001,density_plot=True,y_coordinates=y_coordinates,y_coordinates_2=y_coordinates_2)
```
